# From Hard to Soft Alignments

## Introduction

These educational notebooks accompany the article **From Hard to Soft Alignments:
A Tutorial on Soft Dynamic Time Warping for Music Processing** 

Meinard Müller and Johannes Zeitler, 2026


An audio recording and its symbolic annotation can describe the same content
without sharing a timeline. SDTW uses a differentiable weighting of monotonic
alignment paths to train a model despite these missing local correspondences.
Our running example estimates the activity of four sinusoidal frequency
components directly from a waveform. 

The notebooks assume the background developed in the accompanying article:
DTW paths and local costs, the soft minimum, the global and recursive SDTW
formulations, and gradients as expected alignments. Python, NumPy, basic
PyTorch, and elementary signal processing are useful prerequisites.

## Notebook Overview

| Unit | Notebook | Main topics |
| --- | --- | --- |
| 0 | [Overview](00_overview.ipynb) | Task, learning structure, and setup |
| 1 | [Minimum Functions](01_min_functions.ipynb) | Softmin in 1D and 2D; sparsemin and smoothmin; algebra and stability |
| 2 | [Soft Alignments](02_soft_alignment.ipynb) | Expected paths, temperature, and global versus recursive SDTW |
| 3 | [Strong-Target Training](03_training_strong.ipynb) | Dataset, CNN model, BCE, precision, recall, and F-measure |
| 4 | [Weak-Target Training: Basic](04_training_weak_basic.ipynb) | Unfolded weak targets, batching, modified targets, and temperature scheduling |
| 5 | [Weak-Target Training: Advanced](05_training_weak_advanced.ipynb) | Compact weak targets, step sizes and weights, degenerate alignments, and temporal shifts |

This follows Sections 3–5 of the article: soft minima, SDTW and
its gradient, then learning from strong and weak activity targets.

## Getting Started

From the repository directory, create and activate the environment described
in the README, then open this notebook in Jupyter:

```bash
conda create -n edu_sdtw python=3.12 pip
conda activate edu_sdtw
python -m pip install "ddtw[benchmark,notebook]" scipy
jupyter notebook 00_overview.ipynb
```

`dataset_multipitch.py` provides the synthetic dataset and batching functionality,
`model.py` defines the CNN. The [dDTW toolbox](https://github.com/groupmm/ddtw) 
provides an implementation of the SDTW loss. Our examples run on the CPU by default.
